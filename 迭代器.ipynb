```python
from collections.abc import Iterator
print(list(range(5)))
myiter=iter(list(range(5)))
print(isinstance(myiter,Iterator))
for x in myiter:
    print(x)
```

```python
class MyClass:
    def __init__(self,num):
        self.num=num
    def numlist(self):
        return list(range(self.num))
    def __iter__(self):
        return iter(self.numlist())
myiter=MyClass(num=5)
for x in myiter:
    print(x)
print(isinstance(MyClass,Iterator))
```

```python
#更新MyClass  ,引入next()
class MyClass:
    def __init__(self,num,step,start=0):
        self.num=num
        self.step=step
        self.start=start
    def numlist(self):
        return list(range(self.num))
    def __iter__(self):
        return self
    def __next__(self):
        numlist=self.numlist()
        if self.start<len(numlist):
            numsplit=numlist[self.start:(self.start+self.step)]
            self.start+=self.step
            return numsplit
        else:
            raise StopIteration
myiter=MyClass(num=20,step=5)
for x in myiter:
    print(x)
print(isinstance(myiter,Iterator))
```

「迭代器」的身份必须由 __iter__() 和 __next__() 共同赋予：
__iter__() 确保它能被 for 循环识别为「可提供迭代器的对象」；
__next__() 确保它能实际产生迭代的下一个元素。
缺少任何一个方法，都不能称为迭代器。

```python
import numpy as np
import struct
class MnistDataset:
    def __init__(self,images_file,labels_file):
        self.images_file=images_file
        self.labels_file=labels_file
    def load_images(self,file):
        bin_data=open(file,'rb').read()
        offset=0
        fmt_header='>iiii'
        _,image_num=struct.unpack_from(fmt_header,bin_data,offset)
        offset+=struct.calcsize(fmt_header)
        fmt_image='>B'
        labels=np.empty(image_num)
        for i in range(image_num):
            labels[i]=struct.unpack_from(fmt_image,bin_data,offset)[0]
            offset+=struct.calcsize(fmt_image)
        return labels

    def __getitem__(self.index):
        images=self.load_images(self.images_file)
        labels=self.load_labels(self.labels_file)
        return images[index],labels[index]
if __name__=='__main__':
    images_file='MNIST_data/t10k-images.idx3-ubyte'
    labels_file='MNIST_data/t10k-images.idx1-ubyte'
    dataset=MnistDataset(images_file,labels_file)
    for id,(image,label) in enumerate(dataset):
        print(label)
```

__getitem__ 是 Python 中的一个特殊方法（也叫 “魔术方法”），它的核心作用是让类的实例可以像列表、元组一样通过索引（[]）访问元素。在你的代码中，这个方法为 MnistDataset 类实现了 “按索引获取单个样本（图像 + 标签）” 的功能。
