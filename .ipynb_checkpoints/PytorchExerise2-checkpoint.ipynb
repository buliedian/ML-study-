```python
import torch
import numpy as np
```

```python
#张量的属性
tensor=torch.rand(3,4)
print(tensor.shape)
print(tensor.dtype)
print(tensor.device)
```

```python
from torchvision.models import resnet18,ResNet18_Weights
model=resnet18(weights=ResNet18_Weights.DEFAULT)
data=torch.rand(1,3,64,64)
labels=torch.rand(1,1000)
```

```python
prediction=model(data)
loss=(prediction-labels).sum()
loss.backward()#反向传播开始
```

```python
optim=torch.optim.SGD(model.parameters(),lr=1e-2,momentum=0.9)#加载优化器
```

```python
optim.step()#开始梯度下降
```

```python
a=torch.tensor(1.)
print(type(a))
print(a.grad)
print(a.grad_fn)
```

```python
b=torch.tensor(1.,requires_grad=True)
c=b**2
c.backward()
print(type(b))
print(c.grad)
print(c.grad_fn)
```

```python
a=torch.tensor(1.,requires_grad=True)
b=torch.tensor(2.,requires_grad=True)
a1=torch.tensor(3.,requires_grad=True)
b1=torch.tensor(4.,requires_grad=True)
a2=a*a1
b2=b*b1
c=a2*b2
c.backward()
print(c.grad_fn)
print(c.grad_fn.next_functions)
```
